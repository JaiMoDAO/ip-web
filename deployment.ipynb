## English



1. **Create a Dedicated Repository**
   * Create a new public repository on your GitHub personal or organization root page.
   * **Naming Rule**: The repository name must be strictly formatted as `yourname.github.io` (where `yourname` must match your personal account or organization name exactly).

2. **Import Project Files**
   * Download the complete contents of the `ip-web` repository and upload them to your `yourname.github.io` repository.
   * **Important Note**: **Do not** upload the outer `ip-web` folder itself. Instead, upload all files and subdirectories contained within it, ensuring that `index.html` is located directly in the root directory of the project.

3. **Content Customization & Rendering Optimization**
   * Feel free to modify folders and file contents based on your personal needs.
   * Since `.ipynb` files generally offer better layout and mathematical formula rendering than standard Markdown documents, it is recommended to prioritize using `ipynb` files as the source files for compiling static pages.

4. **Convert `.ipynb` to HTML**
   * Once content modifications are complete, open a new terminal and run the following command to convert the Notebook file into a standard HTML web page:
     ```bash
     jupyter nbconvert --to html OfficialWebsite/PythonTutorials/1PythonChinese.ipynb
     ```

5. **Configure Browser Tab Icons (Favicon)**
   * Open the converted HTML file and locate the `<title>……</title>` tag (typically around line 6).
   * Insert a line break after the closing `</title>` tag and paste the following code to configure the browser tab icon:
     ```html

     <title>……</title>[Insert the following code here.]<script ……></script>

     <!-- 🌟 Set browser tab icon (Favicon) -->
     <link rel="icon" type="image/png" href="../avatar.png">
     <link rel="apple-touch-icon" href="../avatar.png">

     ```
     *(Note: Adjust the `href` path to your actual custom image path based on your directory level)*

6. **Verify and Adjust Link Associations**
   * Go back and open the `index.html` file, then check and update the internal link paths to ensure all redirection links point correctly to your newly created project pages.

7. **Complete Launch and Access**
   * After saving all modifications and pushing them to the remote repository, open `https://yourname.github.io` directly in your browser, and your personal website will be successfully deployed!

## Chinese



1. **创建专属仓库**
   * 在 GitHub 个人或组织根页面创建一个新的公开（Public）项目。
   * **命名规则**：项目名称必须严格命名为 `yourname.github.io`（此处的 `yourname` 必须与你的个人账号或组织名称完全一致）。

2. **导入项目文件**
   * 将 `ip-web` 库的全部内容下载并上传到 `yourname.github.io` 仓库中。
   * **核心注意**：请**切勿**直接上传 `ip-web` 外层文件夹，而是将里面的所有文件和子目录直接上传，确保 `index.html` 文件必须位于项目根目录下。

3. **内容定制与渲染优化**
   * 可根据个人需求自行修改文件夹与文件内容。
   * 项目中由于 `.ipynb` 文件的排版和数学公式渲染效果通常优于普通的 Markdown 文档，建议优先选择 `ipynb` 文件作为静态页面的源文件进行编译。

4. **将 `.ipynb` 转换为 HTML**
   * 确认内容修改完成后，新建终端并运行以下命令，将 Notebook 文件转换为标准的 HTML 网页文件：
     ```bash
     jupyter nbconvert --to html OfficialWebsite/PythonTutorials/1PythonChinese.ipynb
     ```

5. **配置网页标签图标 (Favicon)**
   * 打开转换后的 HTML 文件，找到 `<title>……</title>` 标签（通常在第 6 行左右）。
   * 在 `<title>` 结束标签下方换行，在以下位置插入代码来配置浏览器标签页图标：
     ```html

     <title>……</title>[在此处插入以下代码]<script ……></script>

     <!-- 🌟 设置浏览器标签页图标 (Favicon) -->
     <link rel="icon" type="image/png" href="../avatar.png">
     <link rel="apple-touch-icon" href="../avatar.png">
     
     ```
     *(注：请根据你的实际目录层级，将 `href` 的路径调整为你自行设定的图片地址)*

6. **校对超链接关联**
   * 返回并进入 `index.html` 文件中，重新检查并修改内部的地址关联，确保所有跳转路径正确指向你新建立的专属项目页面。

7. **完成上线与访问**
   * 保存全部修改并推送到远程仓库后，直接在浏览器中打开 `https://yourname.github.io`，你的个人网页即部署成功！
